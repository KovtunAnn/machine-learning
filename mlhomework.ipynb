```python
x = 0.97
y = 8.7
w = 5
lr = 0.1
prediction = w * x
error = (prediction - y) ** 2
grad = 2 * (prediction - y) * x
print("прогноз:", prediction)
print("помилка:", error)
print("крутизна:", grad)
wnew = w - lr * grad
print("нове w:", wnew)
prednew= wnew * x
errornew=(y-prednew)**2
print("новий прогноз:", prednew)
print("нова помилка:", errornew)
print("Стало краще?", errornew < error)
```
